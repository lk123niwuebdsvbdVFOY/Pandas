```
x = data[['age','years_experience','city_tier','performance_score','num_skills','remote_work']]
y = data['annual_salary_usd']
```

```
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25)
```

```
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```
lr_y = lr.predict(x_test)
```

```
from sklearn import metrics
print('mae:',metrics.mean_absolute_error(y_test,lr_y))
print('mse:',metrics.mean_squared_error(y_test,lr_y))
print('Rmse:',np.sqrt(metrics.mean_squared_error(y_test,lr_y)))
print('R2:',metrics.r2_score(y_test,lr_y))
```
