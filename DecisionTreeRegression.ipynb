```
x = data[['age','bmi','children']]
y = data['charges']
```

```
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=30)
```

```
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```
lr_test = lr.predict(x_test)
```

```
from sklearn import metrics
print(metrics.mean_absolute_error(y_test,lr_test))
print(metrics.mean_squared_error(y_test,lr_test))
print(np.sqrt(metrics.mean_squared_error(y_test,lr_test)))
print(metrics.r2_score(y_test,lr_test))
```
