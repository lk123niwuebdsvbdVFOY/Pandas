```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```
data = pd.read_csv("/content/kyphosis.csv")
data.head()# reading a csv file
```

```
plt.bar(data['Age'],data['Number'])
plt.xlabel("Age in months")
plt.ylabel("Number") # polting a bar chart
```

```
plt.scatter(data['Age'],data['Kyphosis'])
```

```
sns.heatmap(data[['Age','Number','Start']])
```

```
sns.scatterplot(data[['Start','Age','Number']])
```

```
x = data[['Age','Number','Start']]
y = data['Kyphosis'] # dvieding data in into x and y varible
```

```
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.3) # spliting the data into the traing and testing
```

```
from sklearn.linear_model import LogisticRegression
lg = LogisticRegression() # crating and traing ml model
lg.fit(x_train,y_train)
```

```
lg_pred = lg.predict(x_test)
lg_pred # testing the model with testing data
```

```
from sklearn.metrics import accuracy_score
accuracy_score(y_test,lg_pred)# finding accuracy pf LogisticRegression
```
