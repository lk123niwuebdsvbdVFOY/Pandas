Broadcasting You might think:

"There is only one number (5). How can NumPy add it to three values?"

NumPy internally broadcasts the value.

It treats it like:

5 5 5

So it actually performs:

10 + 5

20 + 5

30 + 5

Add	a +

Subtract	a - 5

Multiply	a * 2

Divide	a / 10

Array Addition	a + b

Array Multiplication	a * b

Comparison	a > 25

Filtering	a[a > 25]

Broadcasting	a + 5 (5 is applied to every element)
