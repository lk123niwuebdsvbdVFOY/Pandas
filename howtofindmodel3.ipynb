```
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data['sex'] = le.fit_transform(data['sex'])
data['smoker'] = le.fit_transform(data['smoker'])
data['region'] = le.fit_transform(data['region'])
```

```
data.nunique()
```

```
data.head(5)
```

```
x = data[['age','sex','bmi','children','smoker','region']]
y = data['region']
```

```
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25)
```

```
from sklearn.tree import DecisionTreeRegressor
dtr = DecisionTreeRegressor()
dtr.fit(x_train,y_train)
```

```
dtr_y = dtr.predict(x_test)
```

```
from sklearn import metrics
print(metrics.mean_absolute_error(y_test,dtr_y))
print(metrics.mean_squared_error(y_test,dtr_y))
print(np.sqrt(metrics.mean_squared_error(y_test,dtr_y)))
print(metrics.r2_score(y_test,dtr_y))
```
